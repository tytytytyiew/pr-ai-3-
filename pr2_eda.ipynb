## 1. Загрузка реальных данных
Используется UCI Adult Census Income — реальный набор переписных данных. В нём встречаются пропуски, числовые и категориальные признаки. Для загрузки нужен интернет.

```python
adult_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data"
columns = [
    "age", "workclass", "fnlwgt", "education", "education_num", "marital_status",
    "occupation", "relationship", "race", "sex", "capital_gain", "capital_loss",
    "hours_per_week", "native_country", "income"
]
df = pd.read_csv(
    adult_url, header=None, names=columns, na_values="?",
    skipinitialspace=True, comment="|"
)
df = df.dropna(how="all").reset_index(drop=True)
print("Размер:", df.shape)
display(df.head())
```

## 2. Общий осмотр и типы признаков

```python
print(df.info())
print("\\nПропуски:")
display(df.isna().sum().sort_values(ascending=False).to_frame("missing"))
print("\\nСтатистики числовых признаков:")
display(df.describe().T)
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
categorical_cols = df.select_dtypes(exclude=np.number).columns.tolist()
print("Числовые признаки:", numeric_cols)
print("Категориальные признаки:", categorical_cols)
```

## 3. Распределения и выбросы

```python
for col in ["age", "hours_per_week", "capital_gain"]:
    plt.figure(figsize=(7, 4))
    plt.hist(df[col].dropna(), bins=30, edgecolor="black")
    plt.title(f"Распределение: {col}")
    plt.xlabel(col)
    plt.ylabel("Частота")
    plt.tight_layout()
    plt.show()

plt.figure(figsize=(7, 4))
plt.boxplot(df["hours_per_week"].dropna(), vert=False)
plt.title("Boxplot: hours_per_week")
plt.xlabel("Часы работы в неделю")
plt.tight_layout()
plt.show()

df["workclass"].value_counts().head(12).plot(kind="bar", figsize=(9, 4))
plt.title("Наиболее частые категории workclass")
plt.xlabel("Категория")
plt.ylabel("Количество")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## 4. Корреляции и целевая переменная

```python
corr = df[numeric_cols].corr()
plt.figure(figsize=(10, 7))
plt.imshow(corr, aspect="auto", vmin=-1, vmax=1)
plt.colorbar(label="Корреляция")
plt.xticks(range(len(numeric_cols)), numeric_cols, rotation=90)
plt.yticks(range(len(numeric_cols)), numeric_cols)
plt.title("Корреляционная матрица числовых признаков")
plt.tight_layout()
plt.show()

df["income"].value_counts(normalize=True).mul(100).plot(kind="bar", figsize=(6, 4))
plt.title("Распределение целевого класса income")
plt.ylabel("Доля, %")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
