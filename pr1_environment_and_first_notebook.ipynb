## 1. Импорт библиотек и загрузка реального датасета
Используем Wine Dataset из scikit-learn. Это реальные измерения химических характеристик вин, а не синтетические данные.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine

data = load_wine(as_frame=True)
df = data.frame.copy()
df["target_name"] = df["target"].map(dict(enumerate(data.target_names)))

print("Размер таблицы:", df.shape)
print("\\nТипы столбцов:")
display(df.dtypes.to_frame("dtype"))
print("\\nПропуски по столбцам:")
display(df.isna().sum().to_frame("missing"))
print("\\nСтатистики:")
display(df.describe().T)
display(df.head())
```

## 2. Визуальный анализ

```python
numeric_columns = data.data.columns
feature = "alcohol"

plt.figure(figsize=(7, 4))
plt.hist(df[feature], bins=20, edgecolor="black")
plt.title(f"Распределение признака: {feature}")
plt.xlabel(feature)
plt.ylabel("Количество наблюдений")
plt.tight_layout()
plt.show()

df["target_name"].value_counts().plot(kind="bar", figsize=(7, 4))
plt.title("Количество наблюдений по классам вина")
plt.xlabel("Класс")
plt.ylabel("Количество")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
